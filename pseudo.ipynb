Criando pseudononimização no estilo cifra de cézar e anonimização para o nome do usuário

```python
nome='David Lucas Reche Garcia Bento'
chave = len(nome) # valor que sera adicionado a letra para criar a nova letra, EX: A=65, chave =3; 65+3 = 68; 68=D
def pseudonimizar_nomes(dados: str,chave:int) -> str:
    resultado=""
    for c in dados:
        #transforma uma letra em numero, acrescenta o valor da chave e transforma e uma outra letra
        novo_char= chr(ord(c)+chave) # o chr transforma um numero em seu caracter correspondente na tabela ASCII, o ord transforma uma letra em numero, quando acrescenta a "chave" ele esta adicionando uma valor a letra codificada mudando ela para outra, EX: A=65, chave =3; 65+3 = 68; 68=D
        resultado+=novo_char # acrescenta a nova letra a string de decodficação do usuario
    dados=resultado
    return dados 

# converte o dado pseudononimizado em normal novamente
def reverter_pseudo_nomes(dados_pseudo:str,chave:int)->str:
    resultado=""
    for c in dados_pseudo:
        original=chr(ord(c)-chave)
        resultado+=original
    return resultado

#recebe o dado pseudononimizado 
nome = pseudonimizar_nomes(nome,chave)
print(nome)

#recebe a pseudo e converte para nome original 
nome=reverter_pseudo_nomes(nome,chave)
print(nome)

#anonimização de nome, pega a primeira letra do nome e o ultimo nome, troca todos os outros caracteres por "*"
def anonimizar_nome(nome: str) -> str:
    posicao=nome.rindex(" ") # acha o ultimo espaço do nome(o ultimo nome)
    resultado=nome[0].capitalize() # pega a primeira letra do nome do usuario
    # for x in range(posicao): 
    #     resultado+="*"
    resultado+="*****"+nome[posicao+1:].capitalize() #pega todos as letras após a possição do ultimo espaço
    return resultado
print(anonimizar_nome(nome))
    
```

Criando anonimização e pseudononimização para EMAIL


```python
email="David123@gmail.com"

def pseudo_email(email:str) ->str:
    posicao=email.find("@")#encontra a posição do @ dentro da string 
    restoEmail=email[posicao:] # coloca em uma variavel o restante do email apos o @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)+len(restoEmail))#transforma cada caracter da string em seu valor correspondente da tabela ASCII e soma a um valor e em seguida transforma novamente em uma caracter
    resultado+=restoEmail    
    return resultado

original="david123@gmail.com"

original=pseudo_email(email)
print(original)

def reverter_pseudo_email(email:str)->str:
    posicao=email.find("@") # encontra a posição do @ dentro da string
    restoEmail=email[posicao:] # pega o restante dos caracteres depois do @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)-len(restoEmail)) # reverte a transformação dos caracteres do email
    resultado+=restoEmail    
    return resultado

revertido=reverter_pseudo_email(original)
print(revertido)
# metodo para anonimizar email, troca tudo por * menos o que esta depois do @ e o dominio
def anonimizar_email(email: str) -> str:
    letra="@"
    posicao=email.find(letra) # serve para achar a posicao do primeiro @
    resto_email=""
    resultado=""
    if posicao!= -1:
        resto_email=email[posicao-2:]
    for x in range(len(email)-len(resto_email)):
        resultado+="*"
    resultado+=resto_email
    return resultado
email = anonimizar_email(email)
print(email) #anonimizar email, troca tudo por * menos o que esta depois do @ e o dominio
# ex:
#david@gmail.com
#****id@gmail.com
```

Criando anonimização e pseudononimização para os numeros de telefone

```python
# criando variavel de teste para verificar se o metodo funciona
tel= "11913381357"
#Função para pseudononimizar os numeros de telefone, somando os a 
def pseudononimizar_tel(tel:str) -> str:
    resultado=""
    for x in range(len(tel)):
        a=0
        i=tel[x]
        i=int(i)
        i+=len(tel)
        while i>9:
            a=i-9
            i=a
        resultado+=str(i)
    return resultado
tel=pseudononimizar_tel(tel)
print(tel)

# função para trocar os digitos do telefone e soma-los com 5 
def reverter_pseudo_tel(tel:str) -> str:
    resultado=""
    for x in range(len(tel)):
        a=0
        i=tel[x]
        i=int(i)
        i-=len(tel)
        while i<1:
            a=i+9
            i=a
        resultado+=str(i)
    return resultado
tel=reverter_pseudo_tel(tel)
print(tel)

# Função para mostrar apenas os 4 ultimos digitos do telefone de um usuario
def anonimizar_tel(tel:str) -> str:
    resultado=""
    for x in range(len(tel)-4):
        resultado+="*"
    resultado+=tel[-4:] # pega a string e pega os quantro ultimos caracteres para frente
    return resultado
tel=anonimizar_tel(tel)
print(tel)
```

Criando anonimização e pseudononimização de data_cadastro com exeção do ano de contrato 

```python
# Metodo para anoniminizar a data de cadastro de um usuario
dtCadastro='26/09/2023'
def anonimi_dtCadastro(dtcadastro:str) -> str:
    resultado=""
    pocisao= dtCadastro.rindex("/") # rindex serve para achar a ultima posicao do caractere passado, no caso a barra /
    resultado+="**/**" + dtcadastro[pocisao:] #pega os ultimos 4 digitos da data
    return resultado
dtCadastro=anonimi_dtCadastro(dtCadastro)
print(dtCadastro)
```

```python
cpf='176.428.359-73'
#função para criptografar o email do usuario
def pseudo_cpf(cpf:str)->str:
    posicao=cpf.rindex('-')
    resultado=""
    for x in range(len(cpf)-3):
        i=cpf[x]
        if i!='.':
            a=0
            i=int(i)
            i-=len(cpf)
            while i<1:
                a=i+9
                i=a
            resultado+=str(i)
        else:
            resultado+='.'
    resultado+=cpf[posicao:]
    return resultado

cpf=pseudo_cpf(cpf)
print(cpf)
    
def reverter_pseudo_cpf(cpf:str)->str:
    posicao=cpf.rindex('-')
    resultado=""
    for x in range(len(cpf)-3):
        i=cpf[x]
        if i!='.':
            a=0
            i=int(i)
            i+=len(cpf)
            while i>9:
                a=i-9
                i=a
            resultado+=str(i)
        else:
            resultado+='.'
    resultado+=cpf[posicao:]
    return resultado

cpf=reverter_pseudo_cpf(cpf)
print(cpf)

def anonimi_cpf(cpf:str)->str:
    posicao1=cpf.find('.')
    resultado=cpf[:posicao1]
    posicao2=cpf.rindex('-')
    for x in range(len(cpf[posicao1-posicao2])):
        i=cpf[x]
        if i !='.':
            i='*'
        resultado+=i
    resultado+=cpf[posicao2:]
cpf=anonimi_cpf(cpf)
print(cpf)
```
