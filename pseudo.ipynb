NOME USÚARIO

```python
nome='David Lucas Reche Garcia Bento'
chave = len(nome) # valor que sera adicionado a letra para criar a nova letra, EX: A=65, chave =3; 65+3 = 68; 68=D
def pseudonimizar_nomes(dados: str,chave:int) -> str:
    resultado=""
    for c in dados:
        #transforma uma letra em numero, acrescenta o valor da chave e transforma e uma outra letra
        novo_char= chr(ord(c)+chave) # o chr transforma um numero em seu caracter correspondente na tabela ASCII, o ord transforma uma letra em numero, quando acrescenta a "chave" ele esta adicionando uma valor a letra codificada mudando ela para outra, EX: A=65, chave =3; 65+3 = 68; 68=D
        resultado+=novo_char # acrescenta a nova letra a string de decodficação do usuario
    dados=resultado
    return dados 

# converte o dado pseudononimizado em normal novamente
def reverter_pseudo_nomes(dados_pseudo:str,chave:int)->str:
    resultado=""
    for c in dados_pseudo:
        original=chr(ord(c)-chave)
        resultado+=original
    return resultado

#recebe o dado pseudononimizado 
nome = pseudonimizar_nomes(nome,chave)
print(nome)

#recebe a pseudo e converte para nome original 
nome=reverter_pseudo_nomes(nome,chave)
print(nome)

#anonimização de nome, pega a primeira letra do nome e o ultimo nome, troca todos os outros caracteres por "*"
def anonimizar_nome(nome: str) -> str:
    posicao=nome.rindex(" ") # acha o ultimo espaço do nome(o ultimo nome)
    resultado=nome[0].capitalize() # pega a primeira letra do nome do usuario
    # for x in range(posicao): 
    #     resultado+="*"
    resultado+="*****"+nome[posicao+1:].capitalize() #pega todos as letras após a possição do ultimo espaço
    return resultado
print(anonimizar_nome(nome))
    
```

EMAIL USÚARIO

```python
email="David123@gmail.com"

def pseudo_email(email:str) ->str:
    posicao=email.find("@")#encontra a posição do @ dentro da string 
    restoEmail=email[posicao:] # coloca em uma variavel o restante do email apos o @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)+len(restoEmail))#transforma cada caracter da string em seu valor correspondente da tabela ASCII e soma a um valor e em seguida transforma novamente em uma caracter
    resultado+=restoEmail    
    return resultado

# original="david123@gmail.com"

original=pseudo_email(email)
print(original)

def reverter_pseudo_email(email:str)->str:
    posicao=email.find("@") # encontra a posição do @ dentro da string
    restoEmail=email[posicao:] # pega o restante dos caracteres depois do @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)-len(restoEmail)) # reverte a transformação dos caracteres do email
    resultado+=restoEmail    
    return resultado

original=reverter_pseudo_email(original)
print(original)
# metodo para anonimizar email, troca tudo por * menos o que esta depois do @ e o dominio
def anonimizar_email(email: str) -> str:
    letra="@"
    posicao=email.find(letra) # serve para achar a posicao do primeiro @
    resto_email=""
    resultado=""
    if posicao!= -1:
        resto_email=email[posicao-2:]
    for x in range(len(email)-len(resto_email)):
        resultado+="*"
    resultado+=resto_email
    return resultado
email = anonimizar_email(email)
print(email) #anonimizar email, troca tudo por * menos o que esta depois do @ e o dominio
# ex:
#david@gmail.com
#****id@gmail.com
```

TELEFONE USÚARIO

```python
# criando variavel de teste para verificar se o metodo funciona
tel= "+550011913381357" #16
#Função para pseudononimizar os numeros de telefone, somando os a 
def pseudononimizar_tel(tel: str) -> str:
    resultado = "+"
    tel = tel.replace("+", "")
    n = len(tel)
    for digito in tel:
        i = int(digito)
        i = (i + n) % 10  # soma e aplica módulo 10
        resultado += str(i)
    return resultado
tel=pseudononimizar_tel(tel)
print(tel)

# # função para trocar os digitos do telefone e soma-los com 5 
def reverter_pseudo_tel(tel: str) -> str:
    resultado = "+"
    tel = tel.replace("+", "")
    n = len(tel)
    for digito in tel:
        i = int(digito)
        i = (i - n) % 10  # subtrai e aplica módulo 10
        resultado += str(i)
    return resultado

tel=reverter_pseudo_tel(tel)
print(tel)
# Função para mostrar apenas os 4 ultimos digitos do telefone de um usuario
def anonimizar_tel(tel:str) -> str:
    resultado=""
    for x in range(len(tel)-4):
        resultado+="*"
    resultado+=tel[-4:] # pega a string e pega os quantro ultimos caracteres para frente
    return resultado
tel=anonimizar_tel(tel)
print(tel)
```

```python

def pseudonimizar_nomes(dados: str,chave:int) -> str:
    resultado=""
    for c in dados:
        #transforma uma letra em numero, acrescenta o valor da chave e transforma e uma outra letra
        novo_char= chr(ord(c)+chave) # o chr transforma um numero em seu caracter correspondente na tabela ASCII, o ord transforma uma letra em numero, quando acrescenta a "chave" ele esta adicionando uma valor a letra codificada mudando ela para outra, EX: A=65, chave =3; 65+3 = 68; 68=D
        resultado+=novo_char # acrescenta a nova letra a string de decodficação do usuario
    dados=resultado
    return dados 

# converte o dado pseudononimizado em normal novamente
def reverter_pseudo_nomes(dados_pseudo:str,chave:int)->str:
    resultado=""
    for c in dados_pseudo:
        original=chr(ord(c)-chave)
        resultado+=original
    return resultado


#anonimização de nome, pega a primeira letra do nome e o ultimo nome, troca todos os outros caracteres por "*"
def anonimizar_nome(nome: str) -> str:
    posicao=nome.rindex(" ") # acha o ultimo espaço do nome(o ultimo nome)
    resultado=nome[0].capitalize() # pega a primeira letra do nome do usuario
    resultado+="*****"+nome[posicao+1:].capitalize() #pega todos as letras após a possição do ultimo espaço
    return resultado    

def anonimi_senha(senha: str) -> str:
    resultado=""
    for x in senha:
        resultado += "."
    return resultado

def pseudo_email(email:str) ->str:
    posicao=email.find("@")#encontra a posição do @ dentro da string 
    restoEmail=email[posicao:] # coloca em uma variavel o restante do email apos o @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)+len(restoEmail))#transforma cada caracter da string em seu valor correspondente da tabela ASCII e soma a um valor e em seguida transforma novamente em uma caracter
    resultado+=restoEmail    
    return resultado


def reverter_pseudo_email(email:str)->str:
    posicao=email.find("@") # encontra a posição do @ dentro da string
    restoEmail=email[posicao:] # pega o restante dos caracteres depois do @, incluindo o própio
    resultado=""
    for x in range(len(email)-len(restoEmail)):
        i=email[x]
        resultado+=chr(ord(i)-len(restoEmail)) # reverte a transformação dos caracteres do email
    resultado+=restoEmail    
    return resultado


def anonimizar_email(email: str) -> str:
    letra="@"
    posicao=email.find(letra) # serve para achar a posicao do primeiro @
    resto_email=""
    resultado=""
    if posicao!= -1:
        resto_email=email[posicao-2:]
    for x in range(len(email)-len(resto_email)):
        resultado+="*"
    resultado+=resto_email
    return resultado

def pseudononimizar_tel(tel: str) -> str:
    resultado = "+"
    tel = tel.replace("+", "")
    n = len(tel)
    for digito in tel:
        i = int(digito)
        i = (i + n) % 10  # soma e aplica módulo 10
        resultado += str(i)
    return resultado

def reverter_pseudo_tel(tel: str) -> str:
    resultado = "+"
    tel = tel.replace("+", "")
    n = len(tel)
    for digito in tel:
        i = int(digito)
        i = (i - n) % 10  # subtrai e aplica módulo 10
        resultado += str(i)
    return resultado


def anonimizar_tel(tel:str) -> str:
    resultado=""
    for x in range(len(tel)-4):
        resultado+="*"
    resultado+=tel[-4:] # pega a string e pega os quantro ultimos caracteres para frente
    return resultado

def anonimi_cnpj(cnpj: str) -> str:
    cnpj_limpo = cnpj.replace(".", "").replace("/", "").replace("-", "")
    resultado = cnpj_limpo[:2] + "*" * 8 + cnpj_limpo[-4:]
    return f"{resultado[:2]}.{resultado[2:5]}.{resultado[5:8]}/{resultado[8:12]}-{resultado[12:]}"

def anonimi_cpf(cpf:str)->str:
    resultado=cpf[:3]
    for x in range(len(cpf)-5):
        resultado+="*"
    resultado+=cpf[-2:]
    return resultado

def anonimizar_datas(data) -> str:
    if hasattr(data, 'strftime'):
        data_str = data.strftime("%Y-%m-%d")
    else:
        data_str = str(data)  
    pocisao = data_str.find("-") 
    resultado=""
    resultado+=data_str[:pocisao+1] +"**-**"  
    return resultado



def conectar_bd():
    try:
        # Carrega e valida variáveis
        DB_USER = os.getenv("DB_USER")
        DB_PASSWORD = os.getenv("DB_PASSWORD")
        DB_HOST = os.getenv("DB_HOST")
        DB_PORT = os.getenv("DB_PORT")
        DB_DNAME = os.getenv("DB_DNAME")
        
        # Debug: Mostra os valores carregados
        print("=== Variáveis de Ambiente ===")
        print(f"DB_USER: '{DB_USER}'")
        print(f"DB_PASSWORD: {'*' * len(DB_PASSWORD) if DB_PASSWORD else 'VAZIO'}")
        print(f"DB_HOST: '{DB_HOST}'")
        print(f"DB_PORT: '{DB_PORT}'")  # ← Veja se está vazio aqui!
        print(f"DB_DNAME: '{DB_DNAME}'")
        print("============================\n")
        
        if not all([DB_USER, DB_PASSWORD, DB_HOST, DB_PORT, DB_DNAME]):
            print("Erro: Uma ou mais variáveis de ambiente estão vazias!")
            return None
        
        connection_string = f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_DNAME}"
        print(f"Conectando em: postgresql://***:***@{DB_HOST}:{DB_PORT}/{DB_DNAME}\n")
        
        engine = create_engine(connection_string)
        
        with engine.connect() as conn:
            print("Conexão bem-sucedida!\n")
        
        return engine
        
    except Exception as e:
        print(f"Erro ao conectar ao banco de dados: {e}")
        import traceback
        traceback.print_exc()
        return None

#metodo para desconectar do banco de dados  
def desconectar_bd(engine):
    if engine:
        engine.dispose()
        print("Conexão encerrada.")

# função que rodara o código
def processar_dados():
    # conexao com o banco de dados
    engine=conectar_bd() # cria a conexao com o banco de dados

    queryIndustria="""
        SELECT INDUS.ID_INDUSTRIA,INDUS.NOME,INDUS.SENHA,INDUS.CNPJ,INDUS.EMAIL
        FROM INDUSTRIA INDUS
        ORDER BY 1
    """
    queryUsuario="""
        SELECT USE.ID_USUARIO, USE.NOME,USE.SENHA,USE.EMAIL, USE.CPF, USE.DATA_CADASTRO, USE.DATA_NASCIMENTO,USE.ID_INDUSTRIA,
        TEL.ID_TELEFONE, TEL.NUMERO
        FROM USUARIO USE
        JOIN TELEFONE TEL ON USE.ID_USUARIO = TEL.ID_USUARIO
        ORDER BY 1
    """
    # criando os dataframes de usuario e industria
    df_Industria=pd.read_sql(queryIndustria,engine)
    df_Usuario=pd.read_sql(queryUsuario,engine)

    if engine:
        if df_Usuario is not None:
            # NOME
            df_Industria['nome'] = df_Industria['nome']
            df_Usuario['nome_pseudononimizado'] = df_Usuario['nome'].apply(lambda x: pseudonimizar_nomes(x, 5))
            # EMAIL
            df_Usuario['email'] = df_Usuario['email']
            df_Usuario['email_pseudononimizado'] = df_Usuario['email'].apply(anonimizar_email)
            # NUMERO
            df_Usuario['numero'] = df_Usuario['numero']
            df_Usuario['numero_anonimizado'] = df_Usuario['numero'].apply(anonimizar_tel)
            # SENHA
            df_Usuario['senha']=df_Usuario['senha']
            df_Usuario['senha_anonimizado']=df_Usuario['senha'].apply(anonimi_senha)
            # CPF
            df_Usuario['cpf']=df_Usuario['cpf']
            df_Usuario['cpf_anonimizado']=df_Usuario['cpf'].apply(anonimi_cpf)
            # DT_NASCIMENTO
            df_Usuario['data_nascimento'] = df_Usuario['data_nascimento']
            df_Usuario['data_nascimento_anonimizado'] = df_Usuario['data_nascimento'].apply(anonimizar_datas)
            # DT_CADASTRO
            df_Usuario['data_cadastro'] = df_Usuario['data_cadastro']
            df_Usuario['data_cadastro_anonimizado'] = df_Usuario['data_cadastro'].apply(anonimizar_datas)
            
        if df_Industria is not None:
            # NOME
            df_Industria['nome'] = df_Industria['nome']
            df_Industria['nome_pseudononimizado'] = df_Industria['nome'].apply(lambda x: pseudonimizar_nomes(x, 5))
            # EMAIL
            df_Industria['email'] = df_Industria['email']
            df_Industria['email_pseudononimizado'] = df_Industria['email'].apply(pseudo_email)
            # CNPJ
            df_Industria['cnpj'] = df_Industria['cnpj']
            df_Industria['cnpj_anonimizado'] = df_Industria['cnpj'].apply(anonimi_cnpj)
            # SENHA
            df_Industria['senha']=df_Industria['senha']
            df_Industria['senha_anonimizada']=df_Industria['senha'].apply(anonimi_senha)
    
        # ESCREVENDO OS DADOS DO DATAFRAME NOS ARQUIVOS CSV
        df_Industria.to_csv('DADOS_INDUSTRIA.csv', index=False, encoding="utf-8")
        df_Usuario.to_csv('DADOS_USUÁRIO.csv', index=False, encoding="utf-8")
        #desconecta do banco de dados
        desconectar_bd(engine)

processar_dados()
```
